```python
#km to miles

kilometers = float(input("Enter a distance in kilometers: "))
miles = kilometers / 1.609

print(f"{kilometers} kilometers is {miles} miles")
```

```python
#celsius to farenheit

celsius = float(input("Enter a temperature in Celsius: "))
farenheit = (celsius*(9/5)) + 32

print(f"{celsius}°C is {farenheit}°F")
```

```python
#binary to decimal

binary = 101101000
decimal = 0
temp1 = binary
power = 0

for i in range (9):
  digit = temp1%10
  decimal += digit*(2**power)
  power += 1
  temp1 //= 10

print(f"{binary} in binary is the same as {decimal} in decimal")
```

```python
#binary to decimal, takes input (i coded this because i missed the directions saying that it was just 1 number that i had to convert)
#i also found an easier way to do it after i wrote this, but it's already done, so whatever

import math

binary = int(input("Enter a number in binary: "))
decimal = 0
temp1 = binary
power = 0

#i looked up how to get a number's amount of digits and saw math.log10() brought up as a fast way, so i used it :)
numDigits = int(math.log10(abs(binary))) + 1

for i in range (numDigits):
  digit = temp1%10
  decimal += digit*(2**power)
  power += 1
  temp1 //= 10

print(f"{binary} in binary is the same as {decimal} in decimal")
```
